## 3. Membership and removal (10 points)
Complete `remove_guest(guests, name)`. If `name` is present, remove **one** occurrence and return `True`. Otherwise leave the list unchanged and return `False`. The function modifies the supplied list; do not create a replacement list.

```python
def remove_guest(guests, name):
    if name in guests:
      guests.remove(name)
      return True

    else:
      return False

names = ["Ada", "Maya", "Ada", "Leah"]
print(remove_guest(names, "Ada"))   # Expected: True
print(names)                        # Expected: ['Maya', 'Ada', 'Leah']
print(remove_guest(names, "Zoe"))   # Expected: False
```

## 4. Count without using `.count()` (10 points)
Write `count_color(colors, target)` using a `for` loop and an accumulator. Return how many times `target` appears. Handle an empty list.

```python
def count_color(colors, target):
    count = 0

    for color in colors:
      if color == target:
        count += 1

    return count

print(count_color(["red", "blue", "red", "green"], "red")) # 2
print(count_color([], "red")) # 0
```

## 5. Create a filtered list (10 points)
Write `long_names(names, minimum)` that builds and **returns a new list** of names with at least `minimum` characters. Use an empty list, `for`, `if`, and `.append()`. Keep the original list unchanged.

```python
def long_names(names, minimum):
    long = []

    for name in names:
      if len(name) >= minimum:
        long.append(name)

    return long

original = ["Jo", "Amira", "Sam", "Rachel"]
print(long_names(original, 5))  # ['Amira', 'Rachel']
print(original)                # Should still contain all four names
```

## 6. Safely remove items while traversing (15 points)
Write `remove_expired(items, expired)` to remove **all occurrences** of any expired item from `items`. You must use `for item in items.copy()` and `.remove()`; do not use a list comprehension. Return the number removed. Explain in a comment why iterating over `items` directly can skip elements.

```python
def remove_expired(items, expired):
  count = 0

  for item in items.copy():
    if item == expired:
      items.remove(item)
      count += 1

  return count


stock = ["milk", "bread", "milk", "apple", "milk"]
print(remove_expired(stock, "milk"))  # 3
print(stock)                          # ['bread', 'apple']

# If you loop through items and remove one, each item will shift to the left and one will be skipped.
```

## 7. Candy Waves: create a wave (10 points)
Complete `make_wave(number)` so it returns a list containing `number` candy labels. Use `range()` and `.append()`. Each label should be the string `"candy"`. For zero, return an empty list.

```python
def make_wave(number):
    candies = []

    for i in range(number):
      candies.append("candy")

    return candies

print(make_wave(3))  # ['candy', 'candy', 'candy']
print(make_wave(0))  # []
```

## 8. Candy Waves: update inventory (15 points)
Complete `process_catches(active, caught)`. For each name in `caught`, remove **one matching candy** from `active` if it exists; otherwise ignore it. Return the number successfully removed. `active` is changed in place. The `caught` list should not change.

```python
def process_catches(active, caught):
    count = 0

    for name in caught:
      if name in active:
        active.remove(name)
        count += 1

    return count

active = ["pink", "blue", "pink", "green"]
caught = ["pink", "gold", "pink"]
print(process_catches(active, caught))  # 2
print(active)                            # ['blue', 'green']
print(caught)                            # unchanged
```
